# Limpieza

La librería ```pyjanitor```, tiene funciones de utilidad para el procesamiento y limpieza de los DataFrames. 

## Limpieza de columnas

pyjanitor.clean_names, recibe un dataset y retorna uno nuevo con los nombres de las columnas procesados: todos los nombres son retornados en minusculas y agrega underscores a los espacios en blanco

````python
import janitor as jn

dfj = pd.DataFrame({
    "A": [1, None, 0],
    " sales numbers ": [30.0, 20, 50]
})
dfj
````

````python
jn.clean_names(dfj)
````

## Limpieza custom

Si bien pyjanitor es una librería practica, tiene ciertas desventajas como que no limpia los espacios en blanco de inicios y finales de los nombres de las columnas, podemos usar el metodo rename nativo de panas con una función como valor del parametro 

````python
def clean_cols(name):
    return (
        name.strip().lower().replace(" ", "_")
    )

dfj.rename(columns=clean_cols)
````
