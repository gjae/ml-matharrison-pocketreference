# Imputación

Esta técnica consiste en tener alguna herramienta para realizar la predicción de los valores nulos o faltantes, mediante algún tipo de entrenamiento. Scikitlearn provee la clase SimpleImputer en el paquete impute.

- Como regla general, cuando se usa, se debería implementar un pipeline que permita procesar todos los datos posteriores

````python
from sklearn.impute import SimpleImputer

columns = df.select_dtypes(include="number").columns
````

````python
imputer = SimpleImputer(strategy="median")

imputer.fit_transform(df[columns])
````

## SimpleImputer

El transformador SimpleImputer acepta un parametro strategy con valores como:

- median: usa la mediana para llenar los espacios nulos con el valor de la mediana
- mean: usa el promedio como valor __(por defecto)__
- most_frequent: usa el valor más frecuente entre las filas para completar los valores nulo

Si se desea implementar un valor constante, aplicar el parametro strategy en "constant" y usar el parametro fill_value que recibe 
el valor con el que serán completados los valores nulos

### Otros marametros de SimpleImputer

- missing_values: int, float, str, np.nan, None or pandas.NA, default=np.nan. Busca las coincidencias que tengan el tipo de valor indicado en el parametro y será esos a los que se les imputará el valor 


más información https://scikit-learn.org/stable/modules/generated/sklearn.impute.SimpleImputer.html
