# Datos

```python
import pandas as pd

df = pd.DataFrame({
    "a": range(5),
    "b": [-100, -50, 0, 200, 1000]
})

df
```

# Estandarización


Algunos algoritmos tales como SVM, tienden a funcionar mejor cuando los datos se encuentran estandarizados. En teoria, todos los datos deberian estar en un valor medio de 0, con una desviación estandar de 1. Scikitlearn provee el paquete preprocessing con la clase StandardScaler

```python
from sklearn import preprocessing

std = preprocessing.StandardScaler()
```

```python
std.fit_transform(df)
```

```python
std.scale_
```

```python
std.mean_
```

```python
std.var_
```

# Escalado a rango

Este mecanismo implica llevar los valores a una escala en un rango de 0 a 1, inclusives. Se debe tener cuidado con valores outliers

```python
from sklearn import preprocessing

mms = preprocessing.MinMaxScaler()

mms.fit_transform(df)
```
