# Datos

```python
import pandas as pd

df = pd.DataFrame({
    "a": range(5),
    "b": [-100, -50, 0, 200, 1000]
})

df
```

# Estandarización


Algunos algoritmos tales como SVM, tienden a funcionar mejor cuando los datos se encuentran estandarizados. En teoria, todos los datos deberian estar en un valor medio de 0, con una desviación estandar de 1. Scikitlearn provee el paquete preprocessing con la clase StandardScaler

```python
from sklearn import preprocessing

std = preprocessing.StandardScaler()
```

```python
std.fit_transform(df)
```

```python
std.scale_
```

```python
std.mean_
```

```python
std.var_
```

# Escalado a rango

Este mecanismo implica llevar los valores a una escala en un rango de 0 a 1, inclusives. Se debe tener cuidado con valores outliers

```python
from sklearn import preprocessing

mms = preprocessing.MinMaxScaler()

mms.fit_transform(df)
```

# Variables dummies

usados para manejar datos categoricos, existen varias alternativas:
- Pandas: tiene el metodo get_dummies
- Pyjanitor: tiene la funcion expand_columns
- Sklearn: tiene la clase OneHotEncoder

## Pandas

```python
X_cat = pd.DataFrame({
    "names":["George", "Paul"],
    "inst": ["Guitar", "Bass"]
})
```

Implementa drop_first para disminuir las dimensiones. Por defecto, get_dummies implementa 2 variables columnas numericas por cada columna categorica

```python
pd.get_dummies(X_cat)
```

```python
pd.get_dummies(X_cat, drop_first=True)
```

## Pyjanitor

```python
import janitor as jn
```

```python
jn.expand_column(X_cat, column_name="names", sep=",")
```

## OneHotEncoder

en este caso, se debe implementar el encoder en el pipeline de los modelos

```python
from sklearn import preprocessing

oh = preprocessing.OneHotEncoder(drop="first")
oh.fit_transform(X_cat)
```

```python
oh.categories_
```

```python
oh.transform(X_cat).toarray()
```

# LabelEncoder

Ademas de OneHotEncoder, existe el LabelEncoder que funciona transformando la categoria en equivalente numerico, esta impone un ordenamiento que puede que no sea el deseado por lo que hay que tener relativo cuidado. Como contra tenemos que unicamente tiene la capacidad de trabajar una columna a la vez

```python
from sklearn import preprocessing

le = preprocessing.LabelEncoder()

le.fit(X_cat.names)
le.transform(X_cat.names)
```

```python
le.classes_
```
